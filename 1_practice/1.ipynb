```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

```python
good_weld_img = Image.open("input_data/bad_weld_vid644_jpeg_jpg.rf.e6550f4e18eb3c630d35f229de5cf860.jpg")
arr = np.array(good_weld_img)
```

```python
print(f"Shape: {arr.shape}")
print(f"Type: {arr.dtype}")
print(f"min/max: {arr.min()}, {arr.max()}")
print(f"Size, KB: {arr.nbytes / 1024:.1f} KB")
```

```python
print(f"Size = {640 * 640 * 3 * 8 / 8 / 1024:.1f} KB")
```

```python
fig, axes = plt.subplots(1, 4, figsize=(20,5))

channels = ["original", "Reds", "Greens", "Blues"]

for i, (ax, title) in enumerate(zip(axes, channels)):

    if i == 0:
        ax.imshow(arr)
    else:
        ax.imshow(arr[:,:,i-1], cmap=channels[i], vmin=0, vmax=255)

    ax.set_title(title)
    ax.axis('off')

plt.show()
```

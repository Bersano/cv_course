```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
```

```python
arr = np.array(Image.open("input_data/bad_weld_vid644_jpeg_jpg.rf.e6550f4e18eb3c630d35f229de5cf860.jpg"))
```

```python
x_min = 300
x_max = 500

y_min = 220
y_max = 400

roi = arr[y_min:y_max, x_min:x_max]
print(f'shape: {roi.shape}')
```

```python
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(20,5))

ax1.imshow(arr)

rect = patches.Rectangle(
    (x_min, y_min),
    roi.shape[1],
    roi.shape[0],
    linewidth=2,
    edgecolor='r',
    facecolor='none'
)

ax1.add_patch(rect)
ax2.imshow(roi)

ax3.imshow(255 - roi)

plt.show()
```

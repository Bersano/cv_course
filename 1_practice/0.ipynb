```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

```python
good_weld_img = (Image \
    .open("input_data/Good-Welding-images_22_jpeg_jpg.rf.318c9d57367dca722350fe8d63b44591.jpg") \
    .convert("L"))
```

```python
print(good_weld_img.size)
```

```python
good_weld_img = np.array(good_weld_img)
```

```python
good_weld_img.shape
```

```python
resolutions = [4,8,16,32,64,good_weld_img.shape[0]]

fig, axes = plt.subplots(1, 6, figsize=(20,5))

for ax, res in zip(axes, resolutions):
    small = Image \
        .fromarray(good_weld_img) \
        .resize((res, res))

    restored = small.resize((256,256), Image.NEAREST)

    ax.imshow(restored, cmap="gray")
    ax.set_title(f"{res}x{res}")
    ax.axis("off")

plt.show()
```
